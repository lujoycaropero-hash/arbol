El perceptrón es uno de los conceptos básicos de las redes neuronales. Se puede entender como una pequeña unidad que recibe varios datos de entrada, les asigna unos pesos según su importancia y luego genera un resultado. Durante el entrenamiento, esos pesos se van modificando para que la red pueda dar respuestas cada vez más acertadas.

Uno de los problemas del perceptrón simple es que no puede resolver algunos problemas que no se pueden separar fácilmente con una línea. Un ejemplo es el problema XOR. Esto se estudió especialmente en 1969, cuando se evidenciaron las limitaciones de los modelos de una sola capa.

Por eso aparecen los perceptrones multicapa o MLP. A diferencia del perceptrón simple, un MLP tiene varias capas: una capa de entrada, una o más capas ocultas y una capa de salida. Las capas ocultas permiten que la red pueda aprender relaciones más complejas entre los datos.

Para que la red aprenda se utiliza la retropropagación o backpropagation. El proceso comienza cuando los datos pasan por la red y esta genera una respuesta. Después se compara esa respuesta con el resultado correcto para saber qué tanto se equivocó. Ese error se devuelve hacia las capas anteriores y se utiliza para saber qué pesos deben cambiarse.

La imagen muestra este proceso de una forma sencilla: primero se hace el cálculo **hacia adelante**, después se **mide el error**, luego se realiza el **paso inverso** y finalmente se **ajustan los pesos**. Este proceso se repite varias veces para que la red vaya mejorando sus resultados.

También se puede utilizar la **API Sequential** para construir este tipo de redes. Su funcionamiento es sencillo porque permite agregar las capas una después de otra, siguiendo el orden en el que se procesarán los datos. Por ejemplo, se puede colocar una capa de entrada, después varias capas ocultas y finalmente una capa de salida.

Para practicar estos modelos está **Fashion-MNIST**, que es un conjunto de datos formado por imágenes de ropa y diferentes prendas. Las imágenes tienen un tamaño de **28 × 28 píxeles** y están divididas en diferentes categorías, como pantalones, camisetas, zapatos, bolsos, vestidos y otras prendas.

Para que una red neuronal pueda trabajar con estas imágenes, cada imagen de 28 × 28 píxeles se puede convertir en una lista de **784 valores**, ya que 28 × 28 = 784. La red utiliza esos valores para aprender las características de cada tipo de prenda. Después del entrenamiento, puede recibir una imagen nueva y tratar de identificar a qué categoría pertenece.

La **retropropagación** es la parte que permite corregir los pesos durante el aprendizaje, mientras que **Sequential** facilita la construcción de la red y **Fashion-MNIST** sirve como conjunto de datos para probar su funcionamiento.
