 Boosting
Es una técnica de aprendizaje en conjunto (ensemble).

Crea varios clasificadores secuencialmente.

Cada nuevo clasificador intenta corregir los errores cometidos por los anteriores.

La idea es mejorar poco a poco el modelo final.

AdaBoost es un ejemplo: aumenta el peso de los ejemplos que fueron clasificados incorrectamente.
 Ejemplo del 51%
Si un clasificador tiene un 51% de probabilidad de acertar, parece apenas mejor que lanzar una moneda.
Pero al combinar muchos clasificadores independientes, la probabilidad de que la decisión conjunta sea correcta puede aumentar mucho.
 Clasificadores por votación
Se combinan las predicciones de varios modelos:

Hard Voting: gana la clase que recibe más votos.

Simple Majority Vote: se elige la clase con la mayoría simple de votos.

Weighted Voting: algunos clasificadores tienen más peso que otros.

Voting por confianza: se tienen en cuenta las probabilidades/confianzas que entrega cada modelo, no solamente su voto.

 Bagging
Entrena varios modelos en paralelo usando diferentes muestras de los datos.

Después combina sus predicciones.

Busca reducir la varianza y hacer el modelo más estable.

Pasting
Es parecido a Bagging, pero las muestras se toman sin reemplazo.

Cada observación seleccionada no puede aparecer varias veces dentro de la misma muestra.

OOB (Out-of-Bag)
En Bagging, como las muestras se obtienen con reemplazo, algunas observaciones quedan fuera de la muestra usada para entrenar un árbol.

Esas observaciones son las OOB.

Se pueden utilizar para evaluar el modelo sin necesitar necesariamente un conjunto de validación separado.

En una frase
Voting combina modelos votando; Bagging los entrena en paralelo con diferentes muestras; Boosting los entrena secuencialmente para corregir errores; Pasting hace Bagging sin reemplazo; y OOB usa los datos que quedaron fuera para evaluar.
