Reducción de dimensionalidad

Qué es: técnica de Machine Learning que reduce el número de características de un conjunto de datos, conservando la información más relevante posible (por ejemplo, pasar de 100 variables a 2 o 3).

Por qué importa — la maldición de la dimensionalidad:
Cuando hay demasiadas dimensiones, los datos se dispersan en el espacio, lo que trae varios problemas:

Se necesitan más datos para entrenar bien un modelo
Sube el costo computacional
Aumenta el riesgo de sobreajuste (overfitting)
Es más difícil visualizar los datos

Cómo se reduce el espacio:
Proyectando los datos desde un espacio de muchas dimensiones a uno de pocas (ej. 100D → 2D). El Manifold Learning parte de la idea de que, aunque los datos parezcan complejos en muchas dimensiones, en realidad pueden estar organizados sobre una estructura más simple (un "manifold") de menor dimensión.

Principales técnicas:

Técnica	Idea principal
PCA	Encuentra "componentes principales" que conservan la mayor variación posible de los datos
Incremental PCA (IPCA)	Aplica PCA por mini lotes, útil cuando el dataset es muy grande para caber en memoria
LLE	Conserva las relaciones entre cada punto y sus vecinos más cercanos
MDS	Conserva las distancias entre todos los pares de puntos
Isomap	Conserva distancias geodésicas (siguiendo la estructura curva de los datos), útil para datos no lineales

En una frase: muchas dimensiones → datos dispersos y difíciles de manejar → se aplica reducción dimensional → datos más simples de analizar y visualizar, sin perder lo esencial.
